```python
x = np.arange(1_000_000, dtype=float)  

def loop_version(x):
    y = np.empty_like(x) #creates an empty array with the same shape and type as x
    for i in range(len(x)):
        y[i] = x[i] ** 2 #squares each element of x and stores it in y
    return y

def vectorised_version(x):
    return x ** 2 #squares each element of x and returns a new array with the results
```

```python
%timeit loop_version(x) #measures the time it takes to run the loop_version function
%timeit vectorised_version(x) #measures the time it takes to run the vectorized_version function
```

```python
def poly_coeff_arctan_ext(n):
    """The first n coefficients of the arctan series: 0, 1, 0, -1/3, 0, 1/5, 0, -1/7, ..."""
    k = np.arange(n)
    coeffs = np.zeros(n)
    odd = k % 2 == 1
    coeffs[odd] = (-1.0) ** (k[odd] // 2) / k[odd]
    print(coeffs)
    return coeffs

# Evaluating the polynomial at 1 just adds up its coefficients, so a running sum gives the
# approximation of pi for every number of terms at once (no Python loop needed):
errors = np.abs(4 * np.cumsum(poly_coeff_arctan_ext(100_000)) - np.pi)
n_terms = np.argmax(errors < 1e-4) + 1
print("terms needed for an error below 1e-4:", n_terms)

mypi = 4 * Polynomial(poly_coeff_arctan_ext(n_terms))(1)
print(mypi, np.pi)
```
